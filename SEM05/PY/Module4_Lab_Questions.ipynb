## Q1. Student Performance Analysis

A college maintains student records with student name, Mathematics marks, Science marks, Computer Science marks and attendance percentage.

1. Create a Pandas DataFrame
2. Calculate the average marks of each student
3. Identify the student with the highest average
4. Subject-wise average marks as a bar chart
5. Pie chart of students in different grade categories
6. Line chart of one student's marks across subjects
7. Histogram of average marks
8. Scatter plot of attendance vs average marks

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Name": ["Aarav", "Diya", "Kabir", "Meera", "Rohan", "Isha"],
    "Mathematics": [78, 45, 91, 60, 35, 88],
    "Science": [85, 52, 88, 72, 41, 90],
    "ComputerScience": [66, 38, 95, 55, 49, 76],
    "Attendance": [88, 92, 75, 81, 68, 95],
}

df = pd.DataFrame(data)
print(df)
```

```python
# 2. average marks of each student
subjects = ["Mathematics", "Science", "ComputerScience"]
df["Average"] = df[subjects].mean(axis=1)

print(df[["Name", "Average"]].round(2))
```

```python
# 3. student with the highest average
topper = df.loc[df["Average"].idxmax()]
print("Highest average:", topper["Name"], "-", round(topper["Average"], 2))
```

```python
# 4. subject-wise average marks as a bar chart
subject_avg = df[subjects].mean()

subject_avg.plot(kind="bar", color="steelblue")
plt.title("Subject-wise Average Marks")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 5. pie chart of grade categories
def grade(avg):
    if avg >= 85:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 55:
        return "C"
    else:
        return "D"


df["Grade"] = df["Average"].apply(grade)
grade_counts = df["Grade"].value_counts()

plt.pie(grade_counts, labels=grade_counts.index, autopct="%1.1f%%")
plt.title("Percentage of Students by Grade")
plt.tight_layout()
plt.show()

print(grade_counts)
```

```python
# 6. line chart of one student's marks across subjects
student = df[df["Name"] == "Kabir"].iloc[0]

plt.plot(subjects, [student[s] for s in subjects], marker="o", color="green")
plt.title("Marks of " + student["Name"] + " across Subjects")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 7. histogram of average marks
plt.hist(df["Average"], bins=5, color="skyblue", edgecolor="black")
plt.title("Distribution of Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python
# 8. scatter plot of attendance vs average marks
plt.scatter(df["Attendance"], df["Average"], color="purple")
plt.title("Attendance vs Average Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")
plt.tight_layout()
plt.show()
```

## Q2. Employee Data Analysis

Employee details with employee ID, age, experience, salary and department.

1. Create a DataFrame
2. Average salary of employees
3. Employee with the highest salary
4. Department-wise average salary as a bar chart
5. Pie chart of employees per department
6. Salary growth of an employee over five years as a line chart
7. Histogram of salary distribution
8. Scatter plot of experience vs salary
9. Use different colours for different departments

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "EmpID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Age": [25, 32, 45, 29, 38, 51, 27, 41],
    "Experience": [2, 7, 18, 4, 12, 25, 3, 15],
    "Salary": [35000, 55000, 82000, 42000, 68000, 95000, 38000, 72000],
    "Department": ["IT", "HR", "Finance", "IT", "Finance", "HR", "IT", "Finance"],
}

df = pd.DataFrame(data)
print(df)
```

```python
# 2. average salary
print("Average salary:", round(df["Salary"].mean(), 2))

# 3. highest paid employee
top = df.loc[df["Salary"].idxmax()]
print("Highest salary:", top["EmpID"], "-", top["Salary"], "in", top["Department"])
```

```python
# 4. department-wise average salary as a bar chart
dept_avg = df.groupby("Department")["Salary"].mean()

dept_avg.plot(kind="bar", color="teal")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print(dept_avg.round(2))
```

```python
# 5. pie chart of employees per department
dept_counts = df["Department"].value_counts()

plt.pie(dept_counts, labels=dept_counts.index, autopct="%1.1f%%")
plt.title("Percentage of Employees by Department")
plt.tight_layout()
plt.show()
```

```python
# 6. salary growth of one employee over five years
years = [2021, 2022, 2023, 2024, 2025]
salary_growth = [35000, 39000, 44000, 50000, 55000]

plt.plot(years, salary_growth, marker="o", color="darkorange")
plt.title("Salary Growth of Employee 101 over Five Years")
plt.xlabel("Year")
plt.ylabel("Salary")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 7. histogram of salary distribution
plt.hist(df["Salary"], bins=5, color="lightcoral", edgecolor="black")
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python
# 8. scatter plot of experience vs salary, 9. coloured by department
colors = {"IT": "blue", "HR": "green", "Finance": "red"}

for dept in df["Department"].unique():
    subset = df[df["Department"] == dept]
    plt.scatter(subset["Experience"], subset["Salary"],
                color=colors[dept], label=dept)

plt.title("Experience vs Salary by Department")
plt.xlabel("Experience (years)")
plt.ylabel("Salary")
plt.legend()
plt.tight_layout()
plt.show()
```

## Q3. Weather Data Correlation Heatmap

A weather station records temperature, humidity, rainfall, wind speed and atmospheric pressure.

1. Create a DataFrame
2. Calculate the correlation matrix
3. Display the correlation heatmap
4. Identify the two variables with the strongest positive correlation
5. Identify the variables with negative correlation

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Temperature": [30, 32, 34, 29, 27, 31, 35, 33, 28, 26],
    "Humidity": [65, 60, 55, 70, 78, 62, 50, 58, 72, 80],
    "Rainfall": [2, 0, 0, 8, 15, 1, 0, 0, 10, 18],
    "WindSpeed": [12, 15, 18, 10, 8, 14, 20, 16, 9, 7],
    "Pressure": [1012, 1010, 1008, 1014, 1016, 1011, 1007, 1009, 1015, 1017],
}

df = pd.DataFrame(data)
print(df)
```

```python
# 2. correlation matrix
corr = df.corr()
print(corr.round(2))
```

```python
# 3. correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Weather Data Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# 4. strongest positive correlation (excluding self-correlation)
pairs = corr.unstack()
pairs = pairs[pairs < 1.0]        # drop the 1.00 self pairs

strongest = pairs.idxmax()
print("Strongest positive correlation:", strongest, "=", round(pairs.max(), 3))

# 5. negative correlations
negative = pairs[pairs < 0]
print()
print("Negative correlations:")
print(negative.round(3))
```

## Q4. Sales Data Correlation Analysis

Product details with price, quantity sold, discount percentage, advertising expenditure and total sales.

1. Create a DataFrame
2. Correlation between numerical variables
3. Correlation matrix as a heatmap
4. Variables most strongly correlated with total sales
5. Suggest which variables are useful for predicting sales

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Price": [100, 250, 150, 400, 220, 180, 320, 90, 500, 275],
    "Quantity": [50, 30, 45, 15, 35, 40, 22, 60, 10, 28],
    "Discount": [5, 10, 8, 15, 12, 6, 18, 4, 20, 9],
    "Advertising": [2000, 3500, 2500, 5000, 3000, 2200, 4500, 1500, 6000, 3200],
    "TotalSales": [4750, 6750, 6210, 5100, 6776, 6768, 5764, 5184, 4000, 7007],
}

df = pd.DataFrame(data)
print(df)
```

```python
# 2. correlation between numerical variables
corr = df.corr()
print(corr.round(2))
```

```python
# 3. heatmap of the correlation matrix
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="YlGnBu", fmt=".2f")
plt.title("Sales Data Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# 4. variables most strongly correlated with TotalSales
sales_corr = corr["TotalSales"].drop("TotalSales").sort_values(ascending=False)

print("Correlation with TotalSales:")
print(sales_corr.round(3))
print()
print("Strongest predictor:", sales_corr.index[0], "=", round(sales_corr.iloc[0], 3))

# 5. recommendation
print()
print("Useful for predicting sales:")
for var, value in sales_corr.items():
    if abs(value) >= 0.5:
        print(" -", var, "(correlation", round(value, 3), ")")
```

## Q5. Iris Dataset Correlation Heatmap

Load the Iris dataset from `sklearn.datasets`.

1. Load the Iris dataset
2. Convert it into a Pandas DataFrame
3. Calculate the correlation matrix
4. Display an annotated heatmap using `sns.heatmap()`
5. Identify the two most highly correlated features
6. Plot those two features as a scatter plot
7. Use different colours for the three flower classes

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# 1. load the dataset
iris = load_iris()

# 2. convert to a DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["species"] = iris.target
df["species_name"] = df["species"].map(dict(enumerate(iris.target_names)))

print(df.head())
```

```python
# 3. correlation matrix of the four features
features = iris.feature_names
corr = df[features].corr()
print(corr.round(3))
```

```python
# 4. annotated heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Iris Feature Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# 5. two most highly correlated features
pairs = corr.unstack()
pairs = pairs[pairs < 1.0]

strongest = pairs.idxmax()
print("Two most correlated features:", strongest, "=", round(pairs.max(), 3))
```

```python
# 6 and 7. scatter plot of the two features, coloured by class
feature_x = strongest[0]
feature_y = strongest[1]

colors = {"setosa": "red", "versicolor": "green", "virginica": "blue"}

for name in iris.target_names:
    subset = df[df["species_name"] == name]
    plt.scatter(subset[feature_x], subset[feature_y],
                color=colors[name], label=name)

plt.title(feature_x + " vs " + feature_y)
plt.xlabel(feature_x)
plt.ylabel(feature_y)
plt.legend()
plt.tight_layout()
plt.show()
```
