## Q1. Find the second-largest element in a list

Write a Python program to find the second-largest element in a list.

```python
numbers = [45, 67, 32, 89, 56, 89]

unique = list(set(numbers))   # remove duplicates first
unique.sort()

print("List:", numbers)
print("Second largest element:", unique[-2])
```

## Q2. Remove duplicate elements from a list

Write a Python program to remove duplicate elements from a list.

```python
numbers = [1, 2, 2, 3, 4, 4, 5, 5, 6]

unique = []
for n in numbers:
    if n not in unique:
        unique.append(n)

print("Original list:", numbers)
print("List without duplicates:", unique)
```

## Q3. Find the frequency of elements in a list

Write a Python program to find the frequency of each element in a list.

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}
for n in numbers:
    if n in frequency:
        frequency[n] += 1
    else:
        frequency[n] = 1

print("List:", numbers)
print("Frequency of each element:")
for key, value in frequency.items():
    print(key, "->", value)
```

## Q5. Find the common elements in two tuples

Write a Python program to find the common elements in two tuples.

```python
t1 = (1, 2, 3, 4, 5)
t2 = (4, 5, 6, 7, 8)

common = []
for item in t1:
    if item in t2 and item not in common:
        common.append(item)

print("Tuple 1:", t1)
print("Tuple 2:", t2)
print("Common elements:", tuple(common))
```

## Q6. Reverse a list and a tuple without using reverse()

Write a Python program to reverse a list and a tuple without using the built-in `reverse()` method.

```python
lst = [1, 2, 3, 4, 5]

reversed_list = []
for i in range(len(lst) - 1, -1, -1):
    reversed_list.append(lst[i])

print("Original list:", lst)
print("Reversed list:", reversed_list)

t = (10, 20, 30, 40)

reversed_tuple = ()
for i in range(len(t) - 1, -1, -1):
    reversed_tuple = reversed_tuple + (t[i],)

print("Original tuple:", t)
print("Reversed tuple:", reversed_tuple)
```

## Q7. Reverse a given string

Write a Python program to reverse a given string.

```python
text = "Python Programming"

reversed_text = ""
for ch in text:
    reversed_text = ch + reversed_text

print("Original string:", text)
print("Reversed string:", reversed_text)
```

## Q1. Create a dictionary of five students and display all details

Create a dictionary containing the roll number, name and marks of five students. Display all student details.

```python
students = {
    101: {"name": "Aarav", "marks": [78, 85, 66]},
    102: {"name": "Diya", "marks": [45, 52, 38]},
    103: {"name": "Kabir", "marks": [91, 88, 95]},
    104: {"name": "Meera", "marks": [60, 72, 55]},
    105: {"name": "Rohan", "marks": [30, 41, 49]},
}

print("Roll  Name    Marks")
for roll, info in students.items():
    print(roll, " ", info["name"], " ", info["marks"])
```

## Q2. Find the student who scored the highest marks

```python
topper_roll = None
topper_name = None
best_total = -1

for roll, info in students.items():
    total = sum(info["marks"])
    if total > best_total:
        best_total = total
        topper_roll = roll
        topper_name = info["name"]

print("Highest scoring student:", topper_name, "(Roll", topper_roll, ")")
print("Total marks:", best_total)
```

## Q3. Display students who scored more than 50 marks

A student's average of the subject marks is used here as the score.

```python
print("Students with average marks more than 50:")

for roll, info in students.items():
    average = sum(info["marks"]) / len(info["marks"])
    if average > 50:
        print(info["name"], "-", round(average, 2))
```

## Q4. Calculate and display the total and average marks of each student

```python
for roll, info in students.items():
    total = sum(info["marks"])
    average = total / len(info["marks"])
    print(info["name"], "-> Total:", total, " Average:", round(average, 2))
```

## Q1. Menu-driven program to add, display, search and append student records

This program is interactive, so run it in a terminal or answer the prompts when the cell runs.

```python
def add_student():
    roll = input("Enter roll number: ")
    name = input("Enter name: ")
    with open("students.txt", "a") as f:
        f.write(roll + "," + name + "\n")
    print("Record added.")


def display_students():
    try:
        with open("students.txt", "r") as f:
            print(f.read())
    except FileNotFoundError:
        print("No records found.")


def search_student():
    roll = input("Enter roll number to search: ")
    found = False
    try:
        with open("students.txt", "r") as f:
            for line in f:
                parts = line.strip().split(",")
                if parts[0] == roll:
                    print("Found:", line.strip())
                    found = True
        if not found:
            print("Record not found.")
    except FileNotFoundError:
        print("No records found.")


# In a terminal this loop asks the user for input().
# In a notebook, scripted_choices feeds the menu so the cell runs top to bottom.
scripted_choices = ["1", "1", "2", "3", "4"]
scripted_inputs = ["101", "Aarav", "102", "Diya", "102"]


def get_choice(prompt):
    if scripted_choices:
        value = scripted_choices.pop(0)
        print(prompt + value)
        return value
    return input(prompt)


def get_input(prompt):
    if scripted_inputs:
        value = scripted_inputs.pop(0)
        print(prompt + value)
        return value
    return input(prompt)


# point input() at the scripted values so the functions work unchanged
input = get_input

while True:
    print("\n1. Add student")
    print("2. Display students")
    print("3. Search student")
    print("4. Exit")
    choice = get_choice("Enter your choice: ")

    if choice == "1":
        add_student()
    elif choice == "2":
        display_students()
    elif choice == "3":
        search_student()
    elif choice == "4":
        print("Exiting...")
        break
    else:
        print("Invalid choice")
```

## Q2. Store employee details in a file, calculate salary statistics, save a report

Read the records, calculate salary statistics and save the salary report in another file.

```python
# create the employee file first
with open("employees.txt", "w") as f:
    f.write("E101,Ravi,45000\n")
    f.write("E102,Anita,62000\n")
    f.write("E103,Suresh,38000\n")
    f.write("E104,Kavya,75000\n")

salaries = []

with open("employees.txt", "r") as f:
    for line in f:
        parts = line.strip().split(",")
        salaries.append(float(parts[2]))

total = sum(salaries)
average = total / len(salaries)
highest = max(salaries)
lowest = min(salaries)

with open("salary_report.txt", "w") as f:
    f.write("SALARY REPORT\n")
    f.write("Number of employees: " + str(len(salaries)) + "\n")
    f.write("Total salary: " + str(total) + "\n")
    f.write("Average salary: " + str(round(average, 2)) + "\n")
    f.write("Highest salary: " + str(highest) + "\n")
    f.write("Lowest salary: " + str(lowest) + "\n")

print("Report saved to salary_report.txt")
print()
print(open("salary_report.txt").read())
```

## Q3. Add book details, display all books, search for a book, append new records

```python
def add_book(book_id, title, author):
    with open("books.txt", "a") as f:
        f.write(book_id + "," + title + "," + author + "\n")


def display_books():
    print("All books:")
    with open("books.txt", "r") as f:
        for line in f:
            print(line.strip())


def search_book(title):
    with open("books.txt", "r") as f:
        for line in f:
            parts = line.strip().split(",")
            if parts[1].lower() == title.lower():
                print("Book found:", line.strip())
                return
    print("Book not found.")


# add the first books
add_book("B01", "Python Basics", "Ravi Kumar")
add_book("B02", "Data Structures", "Meena Rao")

display_books()

print()
search_book("Data Structures")

# append new book records
add_book("B03", "Statistics for Data Science", "Anil Shah")
print()
display_books()
```

## Q4. Count lines, words, characters, vowels and word frequencies in a text file

```python
with open("sample.txt", "w") as f:
    f.write("Python is easy to learn.\n")
    f.write("Python is powerful and Python is popular.\n")

with open("sample.txt", "r") as f:
    text = f.read()

lines = text.splitlines()
words = text.split()
characters = len(text)

vowels = 0
for ch in text.lower():
    if ch in "aeiou":
        vowels += 1

word_frequency = {}
for word in words:
    word = word.lower().strip(".,")
    if word in word_frequency:
        word_frequency[word] += 1
    else:
        word_frequency[word] = 1

print("Number of lines:", len(lines))
print("Number of words:", len(words))
print("Number of characters:", characters)
print("Number of vowels:", vowels)
print("Word frequencies:")
for word, count in word_frequency.items():
    print(word, "->", count)
```

## Q5. Store deposit and withdrawal transactions in a file and calculate the final balance

```python
with open("transactions.txt", "w") as f:
    f.write("deposit,5000\n")
    f.write("withdraw,1500\n")
    f.write("deposit,2000\n")
    f.write("withdraw,800\n")

balance = 0

with open("transactions.txt", "r") as f:
    for line in f:
        parts = line.strip().split(",")
        transaction_type = parts[0]
        amount = float(parts[1])

        if transaction_type == "deposit":
            balance += amount
        elif transaction_type == "withdraw":
            balance -= amount

        print(transaction_type, amount, "-> balance:", balance)

print("Final balance:", balance)
```

## Q6. Remove all blank lines from a file and store the result in a new file

```python
with open("with_blank_lines.txt", "w") as f:
    f.write("Line one\n")
    f.write("\n")
    f.write("Line two\n")
    f.write("\n")
    f.write("Line three\n")

with open("with_blank_lines.txt", "r") as f:
    lines = f.readlines()

with open("without_blank_lines.txt", "w") as f:
    for line in lines:
        if line.strip() != "":
            f.write(line)

print("Original file:")
print(open("with_blank_lines.txt").read())

print("New file without blank lines:")
print(open("without_blank_lines.txt").read())
```
