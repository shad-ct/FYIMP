# Bayes' Rule

## Q1. Student Department Problem

In a college:
- 60% of students belong to the Science department.
- 40% belong to the Arts department.
- 20% of Science students participate in a programming competition.
- 10% of Arts students participate in the competition.

Find the probability that a student who participates in the competition belongs to the Science department.

```python
# P(Science) and P(Arts)
p_science = 0.60
p_arts = 0.40

# P(participates | department)
p_part_given_science = 0.20
p_part_given_arts = 0.10

# total probability of participating
p_participates = p_science * p_part_given_science + p_arts * p_part_given_arts

# Bayes' rule
p_science_given_part = (p_science * p_part_given_science) / p_participates

print("P(participates) =", p_participates)
print("P(Science | participates) =", round(p_science_given_part, 4))
print("Percentage =", round(p_science_given_part * 100, 2), "%")
```

## Q2. Email Spam Problem

30% of incoming emails are spam. A particular word appears in 80% of spam emails and in 10% of non-spam emails.
If an email contains this word, calculate the probability that the email is spam using Bayes' Rule.

```python
p_spam = 0.30
p_not_spam = 0.70

p_word_given_spam = 0.80
p_word_given_not_spam = 0.10

# total probability of the word appearing
p_word = p_spam * p_word_given_spam + p_not_spam * p_word_given_not_spam

# Bayes' rule
p_spam_given_word = (p_spam * p_word_given_spam) / p_word

print("P(word) =", p_word)
print("P(spam | word) =", round(p_spam_given_word, 4))
print("Percentage =", round(p_spam_given_word * 100, 2), "%")
```

# Binomial Distribution

Formula: `P(X = k) = C(n, k) * p^k * (1-p)^(n-k)`

## Q1. A fair coin is tossed 10 times

Find the probability of:
- exactly 6 heads
- exactly 4 tails
- at least 1 head

```python
n = 10
p = 0.5   # probability of heads


def binomial_probability(n, k, p):
    return comb(n, k) * (p ** k) * ((1 - p) ** (n - k))


# exactly 6 heads
p_6_heads = binomial_probability(n, 6, p)

# exactly 4 tails means exactly 6 heads
p_4_tails = binomial_probability(n, 4, 0.5)

# at least 1 head = 1 - P(no heads)
p_at_least_1_head = 1 - binomial_probability(n, 0, p)

print("P(exactly 6 heads) =", round(p_6_heads, 4))
print("P(exactly 4 tails) =", round(p_4_tails, 4))
print("P(at least 1 head) =", round(p_at_least_1_head, 4))
```

## Q2. Multiple-choice guessing

A multiple-choice question has 4 options with only one correct answer. A student guesses the answers to 8 questions.
Find the probability that the student gets exactly 3 correct answers.

```python
n = 8
p = 1 / 4        # chance of a correct guess
k = 3

p_exactly_3 = comb(n, k) * (p ** k) * ((1 - p) ** (n - k))

print("P(exactly 3 correct) =", round(p_exactly_3, 4))
print("Percentage =", round(p_exactly_3 * 100, 2), "%")
```

## Q3. Defective components

A factory produces electronic components and 5% are defective. If 10 components are selected randomly, find the probability that:
- exactly 2 are defective
- no component is defective
- at least 1 component is defective

```python
n = 10
p = 0.05


def binomial_probability(n, k, p):
    return comb(n, k) * (p ** k) * ((1 - p) ** (n - k))


p_exactly_2 = binomial_probability(n, 2, p)
p_none = binomial_probability(n, 0, p)
p_at_least_1 = 1 - p_none

print("P(exactly 2 defective) =", round(p_exactly_2, 4))
print("P(no defective) =", round(p_none, 4))
print("P(at least 1 defective) =", round(p_at_least_1, 4))
```

# Uniform Distribution

For a uniform distribution between `a` and `b`: `P(x1 < X < x2) = (x2 - x1) / (b - a)`

## Q1. Battery lifetime

The lifetime of a battery is uniformly distributed between 500 and 1,000 hours. Calculate the probability that a randomly selected battery lasts:
- less than 600 hours
- more than 800 hours
- between 600 and 900 hours

```python
a = 500
b = 1000


def uniform_probability(x1, x2, a, b):
    return (x2 - x1) / (b - a)


p_less_600 = uniform_probability(a, 600, a, b)
p_more_800 = uniform_probability(800, b, a, b)
p_between = uniform_probability(600, 900, a, b)

print("P(less than 600 hours) =", round(p_less_600, 4))
print("P(more than 800 hours) =", round(p_more_800, 4))
print("P(between 600 and 900 hours) =", round(p_between, 4))
```

# Poisson Distribution

Formula: `P(X = k) = (e^(-λ) * λ^k) / k!`

## Q1. Call center

A call center receives an average of 4 calls per minute. Find the probability of receiving:
- exactly 2 calls in one minute
- exactly 5 calls in one minute
- no calls in one minute

```python
lam = 4


def poisson_probability(k, lam):
    return (exp(-lam) * (lam ** k)) / factorial(k)


print("P(exactly 2 calls) =", round(poisson_probability(2, lam), 4))
print("P(exactly 5 calls) =", round(poisson_probability(5, lam), 4))
print("P(no calls) =", round(poisson_probability(0, lam), 4))
```

## Q2. Hospital emergency department

A hospital emergency department receives an average of 6 patients per hour. Find the probability that in a randomly selected hour:
- exactly 4 patients arrive
- exactly 8 patients arrive
- no patients arrive

```python
lam = 6


def poisson_probability(k, lam):
    return (exp(-lam) * (lam ** k)) / factorial(k)


print("P(exactly 4 patients) =", round(poisson_probability(4, lam), 4))
print("P(exactly 8 patients) =", round(poisson_probability(8, lam), 4))
print("P(no patients) =", round(poisson_probability(0, lam), 4))
```

## Q3. Website visitors

A website receives an average of 10 visitors per minute. Calculate the probability of receiving:
- exactly 8 visitors in one minute
- more than 10 visitors in one minute
- fewer than 3 visitors in one minute

```python
lam = 10


def poisson_probability(k, lam):
    return (exp(-lam) * (lam ** k)) / factorial(k)


p_exactly_8 = poisson_probability(8, lam)

# more than 10 = 1 - P(0 to 10)
p_more_than_10 = 1 - sum(poisson_probability(k, lam) for k in range(0, 11))

# fewer than 3 = P(0) + P(1) + P(2)
p_fewer_than_3 = sum(poisson_probability(k, lam) for k in range(0, 3))

print("P(exactly 8 visitors) =", round(p_exactly_8, 4))
print("P(more than 10 visitors) =", round(p_more_than_10, 4))
print("P(fewer than 3 visitors) =", round(p_fewer_than_3, 4))
```

# Mean, Variance and Standard Deviation

`mean = sum(x) / n`, `variance = sum((x - mean)^2) / n`, `std = sqrt(variance)`

## Q1. Two students and their test marks

- Student A: 70, 72, 68, 71, 69
- Student B: 55, 85, 65, 90, 70

Calculate the mean, variance and standard deviation for both students.

```python
student_a = [70, 72, 68, 71, 69]
student_b = [55, 85, 65, 90, 70]


def statistics(data):
    n = len(data)
    mean = sum(data) / n
    variance = sum((x - mean) ** 2 for x in data) / n
    std_dev = sqrt(variance)
    return mean, variance, std_dev


for name, data in [("Student A", student_a), ("Student B", student_b)]:
    mean, variance, std_dev = statistics(data)
    print(name)
    print("  Mean =", round(mean, 2))
    print("  Variance =", round(variance, 2))
    print("  Standard deviation =", round(std_dev, 2))
    print()

print("Student A is consistent (low std dev).")
print("Student B is inconsistent (high std dev).")
```

## Q2. Quality control - component diameter

Measurements in millimetres: 10.2, 10.1, 10.3, 10.2, 10.4
Calculate the mean, variance and standard deviation.

```python
diameters = [10.2, 10.1, 10.3, 10.2, 10.4]

n = len(diameters)
mean = sum(diameters) / n
variance = sum((x - mean) ** 2 for x in diameters) / n
std_dev = sqrt(variance)

print("Mean =", round(mean, 3))
print("Variance =", round(variance, 4))
print("Standard deviation =", round(std_dev, 4))
```
