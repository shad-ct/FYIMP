# Sin and cos


``` np.linspace(start, stop, number)```

creates evenly spaced values.

here we multipley with 2*pi -- to get evenly spaced num from 0 to 2pi

````python
import numpy as np 
````

````python

x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x,y1)
plt.plot(x,y2)
````

# Plot types : 

````python
x = np.array([1,2,3,4,5])
y = np.array([3.1,3.9,4.1,4.2,4.7])

plt.figure()
plt.plot(x, y)
plt.title("Line Plot")
plt.show()

plt.figure()
plt.bar(x,y)


plt.figure()
plt.scatter(x,y)

plt.figure()
plt.hist(x)


labels = ["chakka" , "manga" , "thenga", "kundi" , "bruh"]

plt.figure()
plt.pie(x , labels = labels)



plt.figure()
plt.boxplot(x)
````

# 3D plotting 


````python
x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x, y, z)
plt.show()
````
